```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Perceptron
from sklearn.metrics import classification_report
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])

AND = np.array([0, 0, 0, 1])
OR = np.array([0, 1, 1, 1])
XOR = np.array([0, 1, 1, 0])
```

```python
def train_perceptron(X, y, func_name):
    model = Perceptron(max_iter=1000, tol=1e-3, random_state=42)
    model.fit(X, y)
    y_pred = model.predict(X)
    
    print(f"\nClassification Report for {func_name}:")
    print(classification_report(y, y_pred))
    
    sns.heatmap(pd.crosstab(y, y_pred, rownames=['Actual'], colnames=['Predicted']), annot=True, cmap='Blues')
    plt.title(f'Confusion Matrix for {func_name}')
    plt.show()
    
    return model
```

```python
model_and = train_perceptron(X, AND, 'AND')
```

```python
model_or = train_perceptron(X, OR, 'OR')
```

```python
model_xor = train_perceptron(X, XOR, 'XOR')
```
